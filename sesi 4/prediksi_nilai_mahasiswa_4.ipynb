```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

jam_belajar = [[1], [2], [3], [4], [5], [6]]
nilai = [50, 55, 65, 70, 80, 85]
```

```python
model = LinearRegression()
model.fit(jam_belajar, nilai)
```

```python
prediksi = model.predict([[7]])

print("=== Prediksi Nilai Mahasiswa ===")
print("Lama belajar: 7 jam")
print("Prediksi nilai:", round(prediksi[0], 2))
```

```python
plt.scatter(jam_belajar, nilai, label="Data asli")
plt.plot(
    jam_belajar,
    model.predict(jam_belajar),
    color="red",
    label="Garis regresi"
)

plt.xlabel("Jam Belajar")
plt.ylabel("Nilai Mahasiswa")
plt.title("Prediksi Nilai Mahasiswa")
plt.legend()
plt.grid(True)
plt.show()
```
